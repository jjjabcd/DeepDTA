```python
import os
import pandas as pd
```

```python
ROOT_DIR = '/home/rlawlsgurjh/hdd/work/DTA/DeepDTA'
RESULTS_DIR = os.path.join(ROOT_DIR, 'results')
```

```python
TASK_NAME = ["Kd", "Ki"]
MAX_FOLD_NUM = 3
```

```python
def get_results(task_names, max_fold):
    # 모든 결과를 모을 리스트
    all_results = []

    for task in task_names:
        print(f"\n================ {task} Task Results ================")
        task_dfs = []
        
        for fold in range(1, max_fold + 1):
            # 파일 경로 생성: .../results/Kd/fold_1/metrics.csv
            file_path = os.path.join(RESULTS_DIR, task, f'fold_{fold}', 'metrics.csv')
            
            if os.path.exists(file_path):
                # CSV 읽기
                df = pd.read_csv(file_path)
                
                # 결과 확인을 위해 fold 정보 추가 (선택사항)
                df['Fold'] = fold
                
                # 리스트에 추가
                task_dfs.append(df)
                
                print(f"[Fold {fold}] MSE: {df['MSE'].values[0]:.4f}, RMSE: {df['RMSE'].values[0]:.4f}, PCC: {df['PCC'].values[0]:.4f}, CI: {df['CI'].values[0]:.4f}, RM2: {df['RM2'].values[0]:.4f}")
            else:
                print(f"Warning: File not found at {file_path}")

        if task_dfs:
            # 현재 Task의 모든 Fold 데이터를 하나로 합침
            combined_df = pd.concat(task_dfs, ignore_index=True)
            
            # 평균(Mean)과 표준편차(Std) 계산 (Fold 컬럼 제외)
            metrics = ['MSE', 'RMSE', 'PCC', 'CI', 'RM2']
            mean_val = combined_df[metrics].mean()
            std_val = combined_df[metrics].std()

            # 결과 출력 formatting
            print(f"--- {task} Average (over {max_fold} folds) ---")
            
            results_table = pd.DataFrame({
                'Metric': metrics,
                'Mean': mean_val[metrics].map('{:.4f}'.format), # 넷째 자리까지 포맷
                'Std': std_val[metrics].map('{:.4f}'.format)    # 넷째 자리까지 포맷
            })
            
            # 인덱스 깔끔하게 정리하여 출력
            print(results_table.to_string(index=False))
            
        else:
            print(f"No data found for task: {task}")
```

```python
get_results(TASK_NAME, MAX_FOLD_NUM)
```
